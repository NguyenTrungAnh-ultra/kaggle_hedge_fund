```python
print(f"\n--- Categorical Columns ---")
cat_cols = ['code', 'sub_code', 'sub_category', 'horizon']
for col in cat_cols:
    if col in df.columns:
        print(f"{col}: {df[col].nunique()} unique values")

print(f"\n--- Time Series Info ---")
if 'ts_index' in df.columns:
    print(f"ts_index: {df['ts_index'].min()} to {df['ts_index'].max()}")
```

```python
print(f"\n--- Target and Weight ---")
for col in ['y_target', 'weight']:
    if col in df.columns:
        print(f"\n{col} stats:")
        print(df[col].describe())
        print(f"Missing values: {df[col].isna().sum()}")
```
